## Derivación Numérica y Comparación con la Solución Analítica

### Diferencias hacia adelante, hacia atrás y centrada

El objetivo de este ejercicio es comparar la precisión de distintos esquemas de derivación numérica mediante el cálculo de la derivada de una función conocida en el intervalo:

$$
-2 \leq x \leq 2
$$

Considere la función

$$
f(x) = 1 + \frac{1}{2}\tanh(2x)
$$



### 1. Implementación de la función y su derivada exacta **(10 puntos)**

Defina una función en Python:

- `f(x)`: devuelve el valor de
  $$
  f(x) = 1 + \frac{1}{2}\tanh(2x)
  $$
  
Además defina su derivada:

- `df_exacta(x)`: devuelve el valor exacto de la derivada.

```python
import numpy as np
import matplotlib.pyplot as plt

# Definición de la función y su derivada exacta
def f(x):
    '''
    TODO: Complete la función a aquí
    '''
    y= 1+ 0.5*np.tanh(2*x)
    
    return y

def df_exacta(x):
    '''
    TODO: Complete la derivada exacta aquí
    '''
    yp= 1/(np.cosh(2*x)**2)
    
    return yp
```

### 2. Implementación de derivadas numéricas **(30 puntos)**

Implemente las siguientes funciones:
- `derivada_hacia_adelante(f, x, h)`,
- `derivada_hacia_atras(f, x, h)`,
- `derivada_centrada(f, x, h)`.

Las cuales reciben la función a derivas `f`, el punto o nodo de evaluación `x` y el ancho del subintervalo `h`.

```python
# Implementación de esquemas de derivadas numéricas

def derivada_hacia_adelante(f, x, h):
    '''
    TODO: Complete la diferencia hacia adelante aquí
    '''
    return (f(x+h) - f(x))/h

def derivada_hacia_atras(f, x, h):
    '''
    TODO: Complete la diferencia hacia atrás aquí
    '''
    return (f(x) - f(x-h))/h
         
def derivada_centrada(f, x, h):
    '''
    TODO: Complete la diferencia centrada aquí
    '''
    return (f(x+h) - f(x-h))/(2*h)
```

### 3. Evaluación numérica de la derivada **(10 puntos)**

Considere los siguientes números de nodos:

N \in \{16, 32, 64\}
$$

Para cada valor de $N$:

### 4. Visualización de resultados **(0 puntos)**

Para cada valor de $N$, ya tenemos código que crea una figura que compara:

- la derivada exacta \(f'(x)\), usando una línea continua,
- la derivada numérica, usando puntos o marcadores.

En total se genera una gráfica por cada valor de N.

```python
# Parámetros del ejercicio
N_values = [16, 32, 64]
x_range = (-2, 2)

for N in N_values:
    # 3. Construcción de la malla
    x = np.linspace(x_range[0], x_range[1], N)
    h = x[1] - x[0] # Ancho del subintervalo según la malla
    
    # Evaluación de derivadas
    '''
    TODO:
    Evalúe aquí las derivadas aproximadas y exacta.
    Reemplace `None` por las llamadas correctas.
    '''
    y_exacta = df_exacta(x)
    y_adelante = derivada_hacia_adelante(f, x, h)
    y_atras = derivada_hacia_atras(f, x, h)
    y_centrada = derivada_centrada(f, x, h)
    
    # 4. Gráfica de comparación
    plt.figure(figsize=(10, 5))
    plt.plot(x, y_exacta, label='Analítica (Exacta)', color='black', linewidth=2)
    plt.scatter(x, y_adelante, label='Hacia Adelante', marker='o', s=20)
    plt.scatter(x, y_atras, label='Hacia Atrás', marker='s', s=20)
    plt.scatter(x, y_centrada, label='Centrada', marker='^', s=20)
    plt.title(f'Comparación de Derivadas (N = {N})')
    plt.xlabel('x')
    plt.ylabel("f'(x)")
    plt.legend()
    plt.grid(True)
    plt.show()
```

### 5. Análisis de error **(10 puntos)**

Para cada valor de \(N\) y para cada método, calcule el error absoluto:

$$
E_i = \left| f'_{\text{num}}(x_i) - f'_{\text{exacta}}(x_i) \right|
$$

y el error promedio a lo largo de todos los puntos.

Se ofrece ya código que geneera una gráfica comparativa para cada valor de N.

```python
for N in N_values:

    x = np.linspace(x_range[0], x_range[1], N)
    h = x[1] - x[0]
    
    # Evaluación de derivadas
    '''
    TODO:
    Evalúe aquí las derivadas aproximadas y exacta.
    Reemplace `None` por las llamadas correctas.
    '''
    y_exacta = df_exacta(x)
    y_adelante = derivada_hacia_adelante(f, x, h)
    y_atras = derivada_hacia_atras(f, x, h )
    y_centrada = derivada_centrada( f, x, h)
    
    # 5. Análisis de error
    '''
    TODO:
    Evalúe aquí los errores absolutos
    Reemplace `None` por las expresiones correctas usando `np.abs(...)`
    '''   
    err_adelante = np.abs(y_exacta - y_adelante)
    err_atras = np.abs(y_exacta - y_atras)
    err_centrada = np.abs(y_exacta - y_centrada)
    
    print(f"--- Resultados para N = {N} (h ≈ {h:.4f}) ---")
    print(f"Error promedio Adelante: {np.mean(err_adelante):.6e}")
    print(f"Error promedio Atrás:    {np.mean(err_atras):.6e}")
    print(f"Error promedio Centrado: {np.mean(err_centrada):.6e}\n")
    
    # Gráfica de error absoluto
    plt.figure(figsize=(10, 4))
    plt.plot(x, err_adelante, 'o-', label='Error Adelante')
    plt.plot(x, err_atras, 's-', label='Error Atrás')
    plt.plot(x, err_centrada, '^-', label='Error Centrado')
    plt.yscale('log') # Escala logarítmica para apreciar mejor las diferencias
    plt.title(f'Error Absoluto por Método (N = {N})')
    plt.xlabel('x')
    plt.ylabel('Error Absoluto')
    plt.legend()
    plt.grid(True, which="both", ls="-")
    plt.show()
    
```
